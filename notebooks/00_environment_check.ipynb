# Step 1 environment check

This notebook verifies that the minimal local Python environment is ready. It does not load the dataset or contact any URL.

```python
import platform
from importlib.metadata import version

import matplotlib
import numpy as np
import pandas as pd
import sklearn
import tldextract

packages = ["jupyterlab", "matplotlib", "numpy", "pandas", "scikit-learn", "tldextract"]
print("Python:", platform.python_version())
for package in packages:
    print(f"{package}: {version(package)}")
```

```python
extract_offline = tldextract.TLDExtract(suffix_list_urls=())
invented = extract_offline("https://login.example.test/path")
assert invented.domain == "test"
assert invented.subdomain == "login.example"
print("Environment check passed; public-suffix parsing stayed offline.")
```
